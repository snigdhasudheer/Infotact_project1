```python
import pandas as pd

df = pd.read_csv("hillstorm_no_indices.csv")

print("Marketing groups:")
print(df["segment"].value_counts())

print("\nConversion summary:")
print(df.groupby("segment")["conversion"].agg(
    ["count", "sum", "mean"]
))

print("\nVisit summary:")
print(df.groupby("segment")["visit"].agg(
    ["count", "sum", "mean"]
))

print("\nSpending summary:")
print(df.groupby("segment")["spend"].agg(
    ["mean", "median", "max"]
))
```

### Dataset Variables & Limitations

- **Treatment variable:** `segment` — identifies the marketing group.
- **Outcome variable:** `conversion` — indicates whether a customer purchased.
- **Additional outcomes:** `spend` and `visit`.
- **Limitations:** This dataset concerns email marketing, not assigned discount amounts. It cannot directly establish the causal effect of a particular discount.
